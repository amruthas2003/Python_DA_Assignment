## Part 1: NumPy Array Operations

### Question 1: Create a 1D NumPy Array

Create a NumPy array named temperatures_w1 containing Week 1 temperature data.

```python
# Create a 1D NumPy Array for Week 1

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 Temperatures:")
print(temperatures_w1)
```

### Question 2: Inspection and Properties

Inspect the shape, data type, and number of elements in the temperatures_w1 array.

```python
# Inspect Array Properties

print("Shape of Array:", temperatures_w1.shape)

print("Data Type:", temperatures_w1.dtype)

print("Number of Elements:", temperatures_w1.size)
```

### Question 3: Array Operations

Convert temperatures from Celsius to Fahrenheit.

Find the maximum, minimum, and mean temperatures for the week.

```python
# Convert Celsius to Fahrenheit

temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:")
print(temperatures_f)

# Maximum Temperature
print("Maximum Temperature:", np.max(temperatures_w1))

# Minimum Temperature
print("Minimum Temperature:", np.min(temperatures_w1))

# Mean Temperature
print("Mean Temperature:", np.mean(temperatures_w1))
```

### Question 4: Array Slicing and Indexing

Extract the following temperatures:

1. First three days of the week.
2. Weekend temperatures (last two days).
3. Middle three days of the week.

```python
# First three days
print("First three days:", temperatures_w1[:3])

# Weekend (last two days)
print("Weekend temperatures:", temperatures_w1[-2:])

# Middle three days
print("Middle three days:", temperatures_w1[2:5])
```

## Part 2: Pandas Series

### Question 1: Creating Pandas Series

Create a Pandas Series named marks with the following data:

Marks: 95, 92, 89, 85, 80

Custom Indices: Rank1, Rank2, Rank3, Rank4, Rank5

```python
# Create Pandas Series

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("Student Marks:")
print(marks)
```

### Question 2: Indexing and Slicing

1. Access the 1st rank student's mark using integer position.
2. Retrieve top 3 ranks using loc.
3. Retrieve the 3rd rank mark using iloc.
4. Filter ranks where marks are greater than 90 using boolean masking.

```python
# Access 1st rank using integer position
print("1st Rank Mark:", marks.iloc[0])

# Retrieve Top 3 Ranks using loc
print("\nTop 3 Ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

# Retrieve 3rd Rank using iloc
print("\n3rd Rank Mark:", marks.iloc[2])

# Boolean Masking: Marks greater than 90
print("\nMarks Greater than 90:")
print(marks[marks > 90])
```

### Question 3: Manipulating Series

1. Modify Rank1 mark to 100.
2. Remove the entry corresponding to Rank5.
3. Compute CGPA by dividing each mark by 10.

```python
# Modify Rank1 mark to 100
marks['Rank1'] = 100

print("Updated Marks:")
print(marks)

# Remove Rank5
marks = marks.drop('Rank5')

print("\nAfter Removing Rank5:")
print(marks)

# Calculate CGPA
cgpa = marks / 10

print("\nCGPA:")
print(cgpa)
```

## Part 3: Pandas DataFrame

### Question 1: Creating Pandas DataFrame

Create a DataFrame named transactions using the given transaction data.

```python
# Create Pandas DataFrame

transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print("Transaction Data:")
print(transactions)
```

### Question 2: Data Exploration

Explore the transactions DataFrame using head, tail, shape, column names, and data types.

```python
# Display first 5 rows
print("First 5 Rows:")
print(transactions.head())

# Display last 5 rows
print("\nLast 5 Rows:")
print(transactions.tail())

# Display shape
print("\nShape:")
print(transactions.shape)

# Display column names
print("\nColumn Names:")
print(transactions.columns)

# Display data types
print("\nData Types:")
print(transactions.dtypes)
```

```python
# Display ProductCategory and Amount columns

print("ProductCategory and Amount:")
print(transactions[['ProductCategory', 'Amount']])

# Retrieve last 3 columns

print("\nLast 3 Columns:")
print(transactions.iloc[:, -3:])
```

### Data Exploration – Filtering and Grouping

1. Filter Region as North and Amount greater than 200.
2. Find value counts for ProductCategory.
3. Find unique values in Region.
4. Group by Region and calculate mean Amount.

```python
# Filter North Region with Amount greater than 200

print("North Region Transactions above 200:")
print(transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
])


# Value Counts of ProductCategory

print("\nProduct Category Value Counts:")
print(transactions['ProductCategory'].value_counts())


# Unique Values in Region

print("\nUnique Regions:")
print(transactions['Region'].unique())


# Mean Amount by Region

print("\nAverage Amount by Region:")
print(transactions.groupby('Region')['Amount'].mean())
```
